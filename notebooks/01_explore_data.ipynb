```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("="*60)
print("MEDISIGHT - DATA EXPLORATION")
print("="*60)
```

```python
# ==========================================
# 1. DIABETES DATASET
# ==========================================
print("\n📊 EXPLORING DIABETES DATASET")
print("-"*60)
```

```python
# Load dataset
diabetes_df = pd.read_csv('../Dataset/Pima_Indians_Diabetes_Dataset/diabetes.csv')

print(f"\n✓ Dataset loaded: {diabetes_df.shape[0]} rows, {diabetes_df.shape[1]} columns")
print(f"\nColumns: {list(diabetes_df.columns)}")

print("\n📈 Basic Statistics:")
print(diabetes_df.describe())
```

```python
print("\n🎯 Target Distribution:")
print(diabetes_df['Outcome'].value_counts())
print(f"Diabetes cases: {diabetes_df['Outcome'].sum()} ({diabetes_df['Outcome'].mean()*100:.1f}%)")
```

```python
# Check for missing values (zeros in medical data are suspicious)
print("\n⚠️  Zero values (potential missing data):")
zero_counts = (diabetes_df == 0).sum()
print(zero_counts[zero_counts > 0])

print("\n✅ Diabetes dataset exploration complete!")
```
